## 2. 图像基本知识

### 2.1 自定义图像

````python
# 手动创建像素值矩阵，模拟图像的BGR三个通道
b = np.array([
    [0, 127, 255],    # B通道像素值
    [255, 0, 255],    # B通道像素值
    [10, 50, 100]     # B通道像素值
], dtype=np.uint8)  # uint8类型，像素值范围0-255

g = b.T             # G通道：B通道的转置
r = b - g           # R通道：B与G的差值（负值自动截断为0）

# cv.merge()：合并BGR三个通道为彩色图像
show(cv.merge([b, g, r]))
````

### 2.2 三种常见图像相互转换

````python
# cv.imread()读取彩色图片，返回BGR格式的NumPy数组
img = cv.imread('pic/bear500x333.jpg')
show(img)
````

````python
# cv.cvtColor()：颜色空间转换
# cv.COLOR_BGR2GRAY：将BGR彩色图转为灰度图（加权平均法：0.299R + 0.587G + 0.114B）
img_gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
show(img_gray)
````

````python
# cv.threshold()：图像二值化
# 参数：(灰度图, 阈值150, 最大值255, 二值化类型)
# THRESH_BINARY：像素值>150设为255（白色），否则设为0（黑色）
_, img_bin = cv.threshold(img_gray, 150, 255, cv.THRESH_BINARY)
show(img_bin)
````

````python
# cv.imwrite()：保存图像到文件，参数为文件路径和图像数组
cv.imwrite('pic/bear_gray.jpg', img_gray)  # 保存灰度图
cv.imwrite('pic/bear_bin.jpg', img_bin)    # 保存二值化图
````

### 2.2 通道分离与合并

````python
# cv.split()：分离BGR三个通道
# 也可以通过切片获取：b = img[:,:,0], g = img[:,:,1], r = img[:,:,2]
b, g, r = cv.split(img)
# 显示红色通道（R通道），以灰度图形式展示
show(r)
````

````python
# cv.merge()：合并BGR三个通道为彩色图像
img_new = cv.merge([b, g, r])
show(img_new)
````

````python
# 手动实现RGB转灰度图（加权平均法）
# 灰度值 = 0.114*B + 0.587*G + 0.299*R（与cv.cvtColor的COLOR_BGR2GRAY等价）
img_gray2 = 0.114*b + 0.587*g + 0.299*r
show(img_gray2)
````

## 2. 加减乘除

````python
# 读取两张灰度图：背景图和前景图
bg = cv.imread('pic/line500x500.jpg', 0)   # 背景图（灰度模式）
ob = cv.imread('pic/hedgehog500x500.jpg', 0)  # 前景图（灰度模式）

# np.hstack()：水平拼接两张图片并排显示
show(np.hstack([bg, ob]))
````

````python
# 图像相加：两张图片的像素值逐元素相加（图像融合效果）
# bg * 0.5 + ob * 0.5：各取50%权重进行混合
img_add = bg * 0.5 + ob * 0.5
show(img_add)
````

````python
# cv.subtract()：图像相减，逐元素减法（差值会自动截断为0，不会出现负数）
# 从融合图中减去背景，得到前景信息
img_sub = cv.subtract(img_add, bg * 0.5)
show(img_sub)
````

````python
# 图像相乘：用于掩膜（mask）操作
mask = cv.imread('pic/mask500x500.jpg', 0)  # 读取掩膜图（灰度）

show(np.hstack([ob, mask]))  # 并排显示前景图和掩膜
````

````python
# cv.multiply()：图像相乘
# ob/1.0转为浮点数，mask/255归一化到[0,1]，相乘后只保留掩膜为白色的区域
img_mul = cv.multiply(ob/1.0, mask/255)
show(img_mul)
````

````python
# 图像相除：用于去噪等操作
ob_noise = cv.imread('pic/hedgehog_noise_500x500.jpg', 0)  # 带噪声的前景图
show(np.hstack([ob, ob_noise]))  # 并排显示原图和噪声图
````

````python
# cv.divide()：图像相除，+1避免除以0的情况
img_div = cv.divide(ob, ob_noise+1)
show(img_div)
````

````python
# 保存三张图片的拼接结果
cv.imwrite('pic/img_divide.jpg', np.hstack([ob, ob_noise, img_div]))
````

````python
# 实例：图像合成（前景+背景+掩膜）
bg = cv.imread('pic/petal500x500.jpg')    # 背景图（彩色）
ob = cv.imread('pic/hedgehog500x500.jpg')  # 前景图（彩色）
mask = cv.imread('pic/mask500x500.jpg')    # 掩膜图（彩色，实际用作二值掩膜）

# 并排显示背景、前景和掩膜
show(np.hstack([bg, ob, mask]))
````

````python
# 使用掩膜提取前景：前景图乘以归一化的掩膜
# mask/255将掩膜归一化到[0,1]，白色区域保留，黑色区域变为0
ob_select = np.float32(ob/1.0) * np.float32(mask/255.0)
show(np.uint8(ob_select))  # 转回uint8类型显示
````

````python
# 使用反掩膜提取背景：背景图乘以(1-归一化掩膜)
# 1-mask/255将掩膜反转，白色区域变为0，黑色区域保留
bg_select = np.float32(bg/1.0) * np.float32(1-mask/255.0)
show(np.uint8(bg_select))  # 转回uint8类型显示
````

````python
# cv.add()：图像相加，将前景（80%权重）与背景融合
# 0.8使前景稍微变暗，使融合效果更自然
nice = cv.add(ob_select*0.8, bg_select).astype(np.uint8)
show(nice)
````

````python
# 保存合成结果：并排显示背景、前景和合成图
cv.imwrite('pic/image_mix.jpg', np.hstack([bg, ob, nice]))
````
