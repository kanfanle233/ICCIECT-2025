```python
import numpy as np              # NumPy数值计算库
import matplotlib.pyplot as plt  # Matplotlib绑图库
import cv2 as cv                 # OpenCV计算机视觉库
```

```python
def show(img):
    """自定义显示函数：自动判断灰度图或彩色图并正确显示"""
    if img.ndim == 2:  # 灰度图
        plt.imshow(img, cmap='gray')
    else:  # 彩色图，BGR转RGB
        plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
    plt.show()
```

```python
# 读取原始图片（蓝色球体）
img = cv.imread('pic/1.png')
show(img)
```

```python
# 步骤1：水平镜像（左右翻转）
img_H = cv.flip(img, 1)  # flipCode=1表示水平翻转
show(img_H)
```

```python
# 步骤2：将原图与水平镜像相加，得到左右对称的图像
img_2 = img + img_H
show(img_2)
```

```python
# 步骤3：对左右对称图进行垂直镜像（上下翻转）
img_vertical = cv.flip(img_2, 0)  # flipCode=0表示垂直翻转
show(img_vertical)
```

```python
# 步骤4：将上下对称图与原左右对称图相加，得到四个方向都对称的最终结果
result = img_2 + img_vertical
show(result)
```
