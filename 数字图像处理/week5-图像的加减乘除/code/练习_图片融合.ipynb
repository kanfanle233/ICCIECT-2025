# 练习：图片融合

请把三张图片，融合成下面这样图片的样式

```python
import numpy as np              # NumPy数值计算库
import matplotlib.pyplot as plt  # Matplotlib绑图库
import cv2 as cv                 # OpenCV计算机视觉库

def show(img):
    """自定义显示函数：自动判断灰度图或彩色图并正确显示"""
    if img.ndim == 2:  # 灰度图（二维数组）
        plt.imshow(img, cmap='gray')
    else:  # 彩色图（三维数组），BGR转RGB
        plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
    plt.show()
```

```python
# 读取三张图片：背景(花瓣)、前景(刺猬)、掩膜
A = cv.imread('pic/petal500x500.jpg')    # 背景图
B = cv.imread('pic/hedgehog500x500.jpg')  # 前景图
M = cv.imread('pic/mask500x500.jpg')     # 掩膜图（白色区域为目标）

show(np.hstack([A, B, M]))  # 并排显示三张图片
```
