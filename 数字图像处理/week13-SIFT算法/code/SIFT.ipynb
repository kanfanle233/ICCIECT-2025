```python
import cv2                  # OpenCV计算机视觉库
from matplotlib import pyplot as plt  # Matplotlib绑图库

img = cv2.imread('img/cube.png')  # 读取立方体图片
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # 转为灰度图，SIFT基于灰度图提取特征

sift = cv2.SIFT_create()          # 创建SIFT（尺度不变特征变换）检测器
kp = sift.detect(gray, None)      # 在灰度图上检测关键点，返回KeyPoint对象列表
cv2.drawKeypoints(img, kp, img)   # 将检测到的关键点绘制在原图上

# OpenCV读取为BGR格式，matplotlib需要RGB格式，需要转换
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title('SIFT Keypoints')  # 显示标题"SIFT关键点"
plt.axis('off')              # 隐藏坐标轴
plt.show()
```
