# 基于LBPH的人脸识别demo

备注：在运行代码之前，需要先安装opencv-contrib-python
- pip install opencv-contrib-python

## 0.导入库

```python
import cv2                  # OpenCV计算机视觉库
import numpy as np            # NumPy数值计算库
import matplotlib.pyplot as plt  # Matplotlib绑图库
```

```python
def show(img):
    """自定义显示函数：自动判断灰度图或彩色图并正确显示"""
    if img.ndim == 2:  # 灰度图
        plt.imshow(img, cmap='gray')
    else:  # 彩色图，BGR转RGB
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.imshow(img)
    plt.show()
```

## 1.加载训练图像

```python
# 加载训练用的人脸照片，以灰度模式直接读取（IMREAD_GRAYSCALE=0）
images = []
images.append(cv2.imread('img/huge1.jpg', cv2.IMREAD_GRAYSCALE))  # 胡歌照片1
images.append(cv2.imread('img/huge1.jpg', cv2.IMREAD_GRAYSCALE))  # 胡歌照片2
images.append(cv2.imread('img/dengziqi1.jpg', cv2.IMREAD_GRAYSCALE))  # 邓紫棋照片1
images.append(cv2.imread('img/dengziqi1.jpg', cv2.IMREAD_GRAYSCALE))  # 邓紫棋照片2
```

## 2.准备对应标签

```python
# 对应标签：每个训练图像对应一个整数标签，顺序与images列表一一对应
labels = [0, 0, 1, 1]  # 0代表胡歌，1代表邓紫棋

# 标签字典，用于将数字标签转换为可读的中文名称
dic = {0: '胡歌', 1: '邓紫棋', -1: '无法识别'}
```

## 3.创建识别器

```python
# 创建LBPH（Local Binary Pattern Histograms）人脸识别器
# LBPH将图像分块，对每块提取LBP直方图作为特征向量，通过比较直方图距离进行识别
# threshold=80：置信度阈值，低于此值认为是已知人脸，高于则返回-1（无法识别）
recognizer = cv2.face.LBPHFaceRecognizer_create(threshold=80)
```

## 4.训练模型

```python
# 使用训练图像和对应标签训练识别器
recognizer.train(images, np.array(labels))  # labels需转为NumPy数组

# 保存训练好的模型到文件，方便后续加载使用
# recognizer.save('lbph_model.yml')
```

## 5.预测新图像

```python
# 加载待识别的人脸图像（灰度模式）
predict_image = cv2.imread('img/dengziqi3.jpg', cv2.IMREAD_GRAYSCALE)

# predict返回：label=预测的标签编号，confidence=置信度（距离值，越小越匹配）
label, confidence = recognizer.predict(predict_image)
```

## 6.输出结果

```python
# putText：在图像上绘制文字，显示识别结果的人名
# org=(shape[0]-5, shape[1]-5)：文字位置在图像右下角
cv2.putText(img=predict_image, text=dic[label], org=(predict_image.shape[0] - 5, predict_image.shape[1] - 5), 
            fontFace=cv2.FONT_HERSHEY_SIMPLEX, fontScale=0.75, color=(0, 255, 0), thickness=1)
show(predict_image)

# 输出识别结果和置信度
print('这人是：', dic[label])
print('置信度', confidence)  # 置信度越小表示匹配度越高
```
