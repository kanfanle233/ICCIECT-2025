# 基于LBP的人脸检测

## 1.导入库

```python
import cv2 as cv              # OpenCV计算机视觉库
import matplotlib.pyplot as plt  # Matplotlib绑图库
```

## 2.读取图片与灰度化

```python
# 读取含人脸的BMP格式图片
img= cv.imread('img/1.bmp')

# 灰度处理：LBP人脸检测器需要灰度图作为输入
gray = cv.cvtColor(img, code=cv.COLOR_BGR2GRAY)
```

## 3.创建人脸检测识别器

```python
# 创建级联分类器（CascadeClassifier）用于人脸检测
# 使用LBP（Local Binary Pattern）特征的改进版Haar级联模型
# face_detect = cv.CascadeClassifier('haarcascades/haarcascade_frontalface_default.xml')  # Haar特征版本
face_detect = cv.CascadeClassifier("xml_files/lbpcascades/lbpcascade_frontalface_improved.xml")  # LBP特征版本，速度更快
```

## 4.人脸检测

```python
# detectMultiScale：多尺度人脸检测，返回检测到的人脸矩形区域
# scaleFactor=1.3：每次图像缩放比例，值越小检测越精细但越慢
# minNeighbors=4：候选矩形的最小邻居数，值越大误检越少
# minSize=(10,10)：检测的最小人脸尺寸
# face_zone返回格式：[[x, y, w, h], ...]，(x,y)为左上角坐标，w和h为宽高

# face_zone = face_detect.detectMultiScale(img, scaleFactor=1.2, minNeighbors=2, 
#                                          minSize=(10, 10), flags=cv.CASCADE_SCALE_IMAGE)
face_zone = face_detect.detectMultiScale(gray, scaleFactor=1.3, minNeighbors=4, minSize=(10,10),
                                     flags=cv.CASCADE_SCALE_IMAGE)

print('识别人脸的信息：\n', face_zone)
```

## 5.绘制人脸区域

```python
# 在原图上绘制检测到的人脸区域
for x, y, w, h in face_zone:
    # 绘制矩形框住人脸，BGR格式[0,0,255]为红色，线宽2
    cv.rectangle(img, pt1=(x, y), pt2=(x+w, y+h), color=[0,0,255], thickness=2)
    # 绘制圆形辅助标注人脸，圆心为矩形中心，半径为宽度的一半，绿色线宽2
    cv.circle(img, center=(x + w//2, y + h//2), radius=w//2, color=[0,255,0], thickness=2)

# img[:,:,::-1]：将BGR格式翻转为RGB格式供matplotlib显示
plt.imshow(img[:,:,::-1])
plt.show()
```
