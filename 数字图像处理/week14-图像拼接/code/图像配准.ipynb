```python
import cv2                  # OpenCV计算机视觉库
from matplotlib import pyplot as plt  # Matplotlib绑图库

template = cv2.imread('map1.png', 1)  # 读取模板图像（1表示彩色模式）
target   = cv2.imread('map2.png', 1)  # 读取目标图像

# 创建ORB（Oriented FAST and Rotated BRIEF）特征检测器
# nfeatures=500：最多检测500个特征点；scaleFactor=1.2：金字塔缩放因子
# nlevels=8：金字塔层数；patchSize=20：描述符计算的区域大小
orb = cv2.ORB_create(nfeatures=500, scaleFactor=1.2, nlevels=8, edgeThreshold=20,
                     firstLevel=0, WTA_K=2, patchSize=20, fastThreshold=20)

kp1, des1 = orb.detectAndCompute(template, None)  # 检测template的特征点kp1和描述符des1
kp2, des2 = orb.detectAndCompute(target, None)    # 检测target的特征点kp2和描述符des2

# BFMatcher：暴力匹配器，NORM_HAMMING用于ORB描述符的汉明距离，crossCheck=True双向匹配
bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
mathces = bf.match(des1, des2)                    # 对两个描述符集合进行匹配
mathces = sorted(mathces, key=lambda x: x.distance)  # 按匹配距离从小到大排序（距离越小越匹配）

# drawMatches：在两张图之间绘制前40个最佳匹配关系线
result = cv2.drawMatches(template, kp1, target, kp2, mathces[:40], None, flags=2)
plt.imshow(result), plt.show()
```
