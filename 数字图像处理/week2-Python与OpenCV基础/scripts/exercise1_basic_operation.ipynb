## 7. 图片读写及显示

```python
# 导入所需的库
import numpy as np           # 数值计算库
import matplotlib.pyplot as plt  # 绑图库
import cv2 as cv              # OpenCV计算机视觉库
```

```python
# cv.imread()：读取图片，返回BGR格式的NumPy数组
img = cv.imread('pic/bear500x333.jpg')
```

```python
# cv.imwrite()：将图像保存到文件，参数为路径和图像数组
cv.imwrite('test/bear.jpg', img)
```

```python
# 使用OpenCV窗口显示图片
# cv.imshow()：创建窗口显示图片，参数为窗口名和图像数组
# cv.waitKey(0)：无限等待按键
# cv.destroyAllWindows()：关闭所有OpenCV窗口
cv.imshow('bear', img)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# 使用matplotlib直接显示OpenCV读取的图片（颜色会不对，因为BGR顺序不匹配）
plt.imshow(img)
plt.show()
```

```python
# cv.cvtColor()：颜色空间转换，将BGR格式转为RGB格式以正确显示
img2 = cv.cvtColor(img, cv.COLOR_BGR2RGB)
plt.imshow(img2)
plt.show()
```

```python
# 以灰度模式读取图片（第二个参数0表示IMREAD_GRAYSCALE）
img3 = cv.imread('pic/bear500x333.jpg', 0)
# 使用OpenCV窗口显示灰度图片
cv.imshow('bear', img3)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# 使用cmap='gray'正确显示灰度图像
plt.imshow(img3, cmap='gray')
plt.show()
```

```python
# 查看灰度图的维度：ndim=2表示二维（高度x宽度）
img3.ndim
```

```python
# 自定义显示函数：自动判断灰度图或彩色图并正确显示
def show(img):
    if img.ndim == 2:  # 灰度图
        plt.imshow(img, cmap='gray')
    else:  # 彩色图：需要BGR转RGB
        plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
    plt.show()
```

```python
# 使用自定义show函数显示彩色图片
apple = cv.imread('pic/apple.jpg')
show(apple)
```

```python
# 以灰度模式读取图片并显示
gray = cv.imread('pic/apple.jpg', 0)
show(gray)
```

## 9. 简单绘图

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 生成绘图数据：x从2到19，y = 2x + 随机噪声
x = np.arange(2, 20)
y = 2 * x + np.random.randint(5, 20, 18)
```

```python
# plt.plot()绘制折线图，'+-'表示加号标记+实线，color='r'表示红色
plt.plot(x, y, '+-', color='r')
plt.show()
```

```python
# 生成0到1之间100个均匀分布点，计算三种幂函数曲线
x = np.linspace(0, 1, 100)
y1 = np.power(x, 0.5)  # y = x^0.5（开方曲线）
y2 = x                  # y = x（直线）
y3 = np.power(x, 1.5)  # y = x^1.5（幂函数曲线）
```

```python
# 同时显示三条曲线，添加图例、坐标标签、网格和坐标范围
plt.plot(x, y1, label='0.5')  # 指数0.5的曲线
plt.plot(x, y2, label='1.0')  # 指数1.0的直线
plt.plot(x, y3, label='1.5')  # 指数1.5的曲线
plt.legend()          # 显示图例
plt.xlabel('r')       # x轴标签
plt.ylabel('s')       # y轴标签
plt.grid()            # 显示网格
plt.xlim([0, 1])     # x轴范围
plt.ylim([0, 1])     # y轴范围
plt.show()
```

```python
# 生成1000个[0, 101)范围内的随机整数，用于绘制直方图
a = np.random.randint(0, 101, 1000)
```

```python
# 绘制直方图：rwidth=0.9设置柱宽比例，color='g'设置绿色
plt.hist(a, rwidth=0.9, color='g')
plt.show()
```

```python
# 生成自定义区间边界：从-0.5到101，步长为1
# 使用-0.5作为边界可以使整数值的柱子居中显示
bins = np.arange(-0.5, 101, 1)
bins
```

```python
# 使用自定义bins绘制更精细的直方图
plt.hist(a, bins)
plt.show()
```
