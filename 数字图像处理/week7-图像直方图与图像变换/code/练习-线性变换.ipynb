# 练习-线性变换

线性变换的公式是
$$
Db = a * Da + b
$$

- 1.请你将图片进行线性变换，a=1.5，b=10。并且说明这样变换后图像的变化效果是什么？
- 2.将线性变换后的图片，以灰度直方图的形式显示出来。

```python
import cv2 as cv              # OpenCV计算机视觉库
import numpy as np            # NumPy数值计算库
import matplotlib.pyplot as plt  # Matplotlib绑图库
```

```python
def show(img):
    """自定义显示函数：自动判断灰度图或彩色图并正确显示"""
    if img.ndim == 2:  # 灰度图
        plt.imshow(img, cmap='gray')
    else:  # 彩色图，BGR转RGB
        plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
    plt.show()
```

```python
# 读取彩色图片用于线性变换练习
img = cv.imread('pic/parrot500x450.jpg')
show(img)
```
