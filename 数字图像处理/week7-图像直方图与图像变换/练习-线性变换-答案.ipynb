# 练习-线性变换

线性变换的公式是
$$
Db = a * Da + b
$$

- 1.请你将图片进行线性变换，a=1.5，b=10。并且说明这样变换后图像的变化效果是什么？
- 2.将线性变换后的图片，以灰度直方图的形式显示出来。

```python
import cv2 as cv              # OpenCV计算机视觉库
import numpy as np            # NumPy数值计算库
import matplotlib.pyplot as plt  # Matplotlib绑图库
```

```python
def show(img):
    """自定义显示函数：自动判断灰度图或彩色图并正确显示"""
    if img.ndim == 2:  # 灰度图
        plt.imshow(img, cmap='gray')
    else:  # 彩色图，BGR转RGB
        plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
    plt.show()
```

```python
# 读取彩色图片用于线性变换练习
img = cv.imread('pic/parrot500x450.jpg')
show(img)
```

```python
# 绘制原图的直方图，展平后统计所有通道的像素值分布
plt.hist(img.ravel(), bins=256, range=[0, 256])
plt.show()
```

1.线性变换

```python
# cv.convertScaleAbs()：线性变换，alpha=1.5（斜率），beta=10（截距）
# s = 1.5 * r + 10，使图像整体变亮、对比度增加
img2 = cv.convertScaleAbs(img, alpha=1.5, beta=10)
show(img2)
```

2.变化前后的直方图

```python
# 绘制线性变换后的直方图，观察像素值分布的变化
plt.hist(img2.ravel(), bins=256, range=[0, 256])
plt.show()
```

## 写法2

```python
# 手动实现线性变换（写法2）：先转为int32避免溢出，再用clip截断到[0,255]
a = 1.5
b = 20
img3 = a * img.astype(np.int32) + b   # 转为int32避免uint8溢出
img3 = np.clip(img3, 0, 255)          # 截断到[0,255]
show(np.uint8(img3))                  # 转为uint8显示
print(img3.max())
```
