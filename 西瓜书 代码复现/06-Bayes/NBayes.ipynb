## 朴素贝叶斯

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()  # 设置 seaborn 默认样式，美化图表
```

### 生成随机数据

```python
from sklearn.datasets import make_blobs
# make_blobs：生成符合高斯分布的聚类数据集
# 朴素贝叶斯假设每个特征在给定类别下服从高斯分布，因此用高斯分布数据来演示
# n_samples：样本数量；n_features：特征维度；centers：簇的数量
# cluster_std：簇内标准差（越大数据越分散）；random_state：随机种子保证可复现
X, y = make_blobs(n_samples = 100, n_features=2, centers=2, random_state=2, cluster_std=1.5) 
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='RdBu')  # 按类别着色
plt.show()
```

```python
# 高斯朴素贝叶斯分类器
# 原理：假设每个特征在给定类别下服从正态分布 N(mu_c, sigma_c^2)
# 后验概率：P(c|x) ∝ P(c) * prod(P(x_i|c))，其中 P(x_i|c) 为高斯概率密度函数
# 训练过程：只需计算每个类别的先验概率 P(c) 和每个特征的均值、方差
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(X, y)  # 训练模型（实际上只是统计各类别的均值和方差）

# 生成大规模测试数据，观察模型在二维平面上的决策边界
rng = np.random.RandomState(0)
X_test = [-6, -14] + [14, 18] * rng.rand(2000, 2)  # 在较大范围内随机生成2000个测试点
y_pred = model.predict(X_test)

# 可视化：深色大点为训练集，浅色小点为测试集预测结果
# 决策边界清晰可见：两种颜色的分界线即为贝叶斯决策面
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='RdBu')
lim = plt.axis()
plt.scatter(X_test[:, 0], X_test[:, 1], c=y_pred, s=20, cmap='RdBu', alpha=0.1)
plt.axis(lim)
plt.show()
```

```python
# predict_proba 返回每个测试样本属于各类别的后验概率
# 这是朴素贝叶斯的优势之一：不仅给出分类结果，还给出概率估计
# 概率越接近 1.0，模型对该分类越确信；接近 0.5 表示不确定
yprob = model.predict_proba(X_test)
yprob[-8:].round(2)  # 显示最后8个样本的分类概率（保留2位小数）
```
