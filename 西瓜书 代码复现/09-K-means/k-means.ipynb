```python
from sklearn.cluster import KMeans  # K-Means 聚类算法
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
from sklearn import datasets
%matplotlib inline
```

```python
# 生成100个二维随机数据点，观察聚类前的数据分布
# K-Means 的目标：将这些点划分为 K 个簇，使得簇内平方和（SSE）最小
X = np.random.rand(100, 2)
plt.scatter(X[:, 0], X[:, 1], marker='o')
```

```python
# 手动实现质心初始化函数
# K-Means 算法步骤：
# 1. 随机选择 K 个数据点作为初始质心（centroid）
# 2. 将每个样本分配到距离最近的质心所在的簇
# 3. 重新计算每个簇的质心（簇内所有点的均值）
# 4. 重复步骤2-3，直到质心不再变化（收敛）
def InitCentroids(X, k):
    # 从数据集中随机选取 k 个索引作为初始质心
    index = np.random.randint(0,len(X)-1,k)
    return X[index]
```

```python
# 使用 scikit-learn 的 KMeans 进行聚类
# n_clusters=2：将数据划分为2个簇
# 算法使用欧氏距离（L2距离）衡量样本与质心的距离
# K-Means 优化目标：min sum(||x_i - mu_c_i||^2)，即最小化簇内平方误差和
kmeans = KMeans(n_clusters=2).fit(X)
label_pred = kmeans.labels_  # 获取每个样本的簇标签（0或1）
# 按簇标签着色，直观展示聚类结果
plt.scatter(X[:, 0], X[:, 1], c=label_pred)
plt.show()
```
