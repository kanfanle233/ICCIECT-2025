## **任务 2：数据的基本统计描述**

编写代码完成以下任务：  
1.加载鸢尾花数据集（Iris Dataset） 并创建 Pandas DataFrame。  
2.使用 describe() 输出 均值（Mean）、中位数（Median）、众数（Mode）、标准差（Std）、方差（Variance）、最大/最小值（Max/Min）。  
3.分别使用 Pandas 方法 单独计算并输出：  
    均值（df.mean()）
    中位数（df.median()）
    众数（df.mode()）
    标准差（df.std()）
    方差（df.var()）
    最大值/最小值（df.max() / df.min()）

```python
# 任务2：数据的基本统计描述
import pandas as pd
import seaborn as sns

# 1. 加载鸢尾花数据集（Iris Dataset）并创建 Pandas DataFrame
iris_df = sns.load_dataset('iris')
print(iris_df.head())

# 2. 使用 describe() 输出 均值（Mean）、标准差（Std）、最小值、四分位数、最大值
# describe() 自动生成数据的统计摘要，包含均值、标准差、最小值、25%/50%/75%分位数、最大值
print(iris_df.describe())

# 均值（Mean）：每列数值的算术平均值，反映数据的集中趋势
mean_values = iris_df.mean(numeric_only=True) # numeric_only=True 避免对非数值列计算均值
print(mean_values)

# 中位数（Median）：将数据排序后位于中间位置的值，不受极端值影响
median_values = iris_df.median(numeric_only=True) # numeric_only=True 避免对非数值列计算中位数
print(median_values)

# 众数（Mode）：数据中出现频率最高的值，可能有多个众数
mode_values = iris_df.mode()
print(mode_values)

# 标准差（Standard Deviation）：衡量数据的离散程度，标准差越大说明数据越分散
std_values = iris_df.std(numeric_only=True) # numeric_only=True 避免对非数值列计算标准差
print(std_values)

# 方差（Variance）：标准差的平方，同样衡量数据的离散程度
variance_values = iris_df.var(numeric_only=True) # numeric_only=True 避免对非数值列计算方差
print(variance_values)

# 最大值（Maximum）：每列数值中的最大值
max_values = iris_df.max(numeric_only=True) # numeric_only=True 避免对非数值列计算最大值
print(max_values)

# 最小值（Minimum）：每列数值中的最小值
min_values = iris_df.min(numeric_only=True) # numeric_only=True 避免对非数值列计算最小值
print(min_values)
```

## 任务 4：数据对象的相似性度量

编写代码计算以下相似性度量：  
1.欧氏距离（Euclidean Distance）  
2.曼哈顿距离（Manhattan Distance）  
3.余弦相似度（Cosine Similarity）  

```python
# 任务4：数据对象的相似性度量
import numpy as np
from scipy.spatial.distance import euclidean, cityblock, cosine

# 示例数据点 (向量) - 假设我们有两个数据点，例如描述两个用户的特征向量
point1 = np.array([1, 2, 3, 4, 5])  # 例如，用户1对 5 个电影的评分
point2 = np.array([5, 4, 3, 2, 1])  # 例如，用户2对这 5 个电影的评分

print("数据点 1:", point1)
print("数据点 2:", point2)
print("\n--- 相似性度量计算结果 ---")

# 1. 欧氏距离 (Euclidean Distance)
# 两点之间的直线距离，是最常用的距离度量方式
# 公式：d = sqrt(sum((x_i - y_i)^2))
euclidean_distance_value = euclidean(point1, point2)
print("欧氏距离:", euclidean_distance_value)

# 2. 曼哈顿距离 (Manhattan Distance)
# 两点在各维度上差值的绝对值之和，如同在城市街区中沿网格行走的距离
# 公式：d = sum(|x_i - y_i|)
manhattan_distance_value = cityblock(point1, point2)
print("曼哈顿距离:", manhattan_distance_value)

# 3. 余弦相似度 (Cosine Similarity)
# 衡量两个向量方向的相似程度，取值范围[-1, 1]，值越接近1越相似
# 注意：scipy.spatial.distance.cosine 返回的是余弦距离（1-余弦相似度），需要用1减去才是余弦相似度
cosine_similarity_value = 1 - cosine(point1, point2)
print("余弦相似度:", cosine_similarity_value)
```
