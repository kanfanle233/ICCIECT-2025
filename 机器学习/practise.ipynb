```python
import pandas as pd
import numpy as np

# 示例数据：NumPy数组
data = np.array([['Alice', 25, 'Female'],
                 ['Bob', 30, 'Male'],
                 ['Charlie', 22, 'Male']])

# 列名
columns = ['姓名', '年龄', '性别']

# 从NumPy数组创建DataFrame
df = pd.DataFrame(data, columns=columns)

# 打印DataFrame
print(df)
```

```python
import pandas as pd

# 创建一个示例 DataFrame
data = {'列1': [1, 2, 3],
        '列2': [4.0, 5.0, 6.0],
        '列3': ['a', 'b', 'c']}
df = pd.DataFrame(data)

print("原始 DataFrame:")
print(df)
print("\nDataFrame 的类型:", type(df)) # 打印 DataFrame 的类型

# 使用 to_numpy() 方法转换为 NumPy 数组
numpy_array = df.to_numpy()

print("\n转换后的 NumPy 数组:")
print(numpy_array)
print("\nNumPy 数组的类型:", type(numpy_array)) # 打印 NumPy 数组的类型
```

```python
import pandas as pd

data = {'列1': [1, 2, 3],
        '列2': [4.0, 5.0, 6.0]}
df = pd.DataFrame(data)

dict_data = df.to_dict() # 默认 orient='dict'

print("使用 to_dict() 转换后的 Python 字典 (orient='dict'):")
print(dict_data)
print("\n字典的类型:", type(dict_data))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
data =  {'feature1': range(10), 'feature2': range(10, 20)}
df = pd.DataFrame(data)
X = df[['feature1','feature2']]
y = pd.Series(range(100, 110))

print("--- 不设置 random_state (每次运行结果不同) ---")
for i in range(2):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
print(f"运行 {i+1}: 训练集 X_train 的索引: {X_train.index.tolist()}")

print("\n--- 设置 random_state=42 (每次运行结果相同) ---")
for i in range(2): # 运行两次
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42) # 设置 random_state=42
    print(f"运行 {i+1}: 训练集 X_train 的索引: {X_train.index.tolist()}") 
```
