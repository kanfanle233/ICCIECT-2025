### 通常由两部分组成：函数头（包含函数名和参数列表）、函数体（函数实际执行的代码）

### 如何定义函数

```python
# 语法格式：使用def关键字来定义函数
def function_name(parameters):
    # Function body
    # Some code to execute
    
# 示例
    def greet(name):
        print(f"Hello, {name}!")
```

### 函数的参数传递

```python
# 位置参数
def add(a, b):
    return a + b

result = add(5, 3)  # 调用函数并传递位置参数
print(result)  # 输出 8
```

```python
# 关键字参数
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

greet(age=25, name="Alice")  # 关键字参数，顺序不重要
```

```python
# 默认参数值
def greet(name, age=18):
    print(f"Hello {name}, you are {age} years old.")

greet("Bob")  # 使用默认的 age 值 18
greet("Alice", 22)  # 传递自定义的 age 值 22
```

```python
# 函数没有返回值，也就是没有使用return语句，或者return后没有指定值，函数会返回None
def greet(name):
    print(f"Hello, {name}!")

result = greet("Charlie")  # 输出：Hello, Charlie!
print(result)  # 输出：None
```

# 3. Numpy数值运算基础
## 3.1Numpy简介：介绍Numpy数组（ndarry），与Python列表的区别和优势
### Numpy 数组（ndarray）是 Numpy 的核心数据结构，它比 Python 原生的列表具有更高的性能和更强的功能。
## Numpy与Python列表区别：
### 1️⃣Python 列表是一个通用容器，可以存储不同类型的数据。而 Numpy 数组（ndarray）是一个 同质 数据容器，即数组中的所有元素必须具有相同的类型（通常是数字）。
### 2️⃣Numpy 数组支持矢量化运算，而 Python 列表不支持。矢量化运算能够显著提高计算效率。
### 3️⃣Numpy 数组提供了大量的内建数学函数，可以对数组进行批量运算，而 Python 列表需要使用循环来进行逐个运算。

```python
import numpy as np

# 创建一个 Numpy 数组
arr = np.array([1, 2, 3, 4, 5])
print(arr)  # 输出：[1 2 3 4 5]
```

```python
# 数组创建 np.array()可以从Python列表或元组创建Numpy数组
arr = np.array([1, 2, 3, 4])
print(arr)  # 输出：[1 2 3 4]

# np.zeros():创建一个指定形状的数组，数组中的元素全部为零
# np.ones():创建一个指定形状的数组，数组中的元组全部为一
arr_zeros = np.zeros((3, 3))  # 3x3 的零矩阵
print(arr_zeros)
# 输出：
# [[0. 0. 0.]
#  [0. 0. 0.]
#  [0. 0. 0.]]

arr_ones = np.ones((2, 4))  # 2x4 的一矩阵
print(arr_ones)
# 输出：
# [[1. 1. 1. 1.]
#  [1. 1. 1. 1.]]

# np.arange():返回一个等差数列，类似于Python内建的range()函数
arr_range = np.arange(0, 10, 2)  # 从 0 到 10，步长为 2
print(arr_range)  # 输出：[0 2 4 6 8]

# 数组索引：Numpy 数组可以通过整数索引来访问数组中的元素
arr = np.array([1, 2, 3, 4, 5])
print(arr[0])  # 输出：1
print(arr[2])  # 输出：3

# 数组切片：类似于 Python 列表，Numpy 数组也支持切片操作。
arr = np.array([1, 2, 3, 4, 5])
print(arr[1:4])  # 输出：[2 3 4]
print(arr[:3])   # 输出：[1 2 3]
print(arr[::2])  # 输出：[1 3 5]

# 数组之间的加、减、乘、除：1.Numpy 支持直接对数组进行数学运算，无需使用循环；2.这些运算会自动应用到数组中的每个元素上（称为“矢量化”）。
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print(arr1 + arr2)  # 输出：[5 7 9]
print(arr1 - arr2)  # 输出：[-3 -3 -3]
print(arr1 * arr2)  # 输出：[4 10 18]
print(arr1 / arr2)  # 输出：[0.25 0.4 0.5]

# 广播（Broadcasting）
# 广播是一种强大的机制，它允许不同形状的数组之间进行数学运算。Numpy 会自动将小数组扩展到大数组的形状。
# 广播规则：如果两个数组的形状不匹配，Numpy 会尝试通过扩展维度较小的数组来使它们的形状匹配。
arr1 = np.array([1, 2, 3])
arr2 = np.array([10])

print(arr1 + arr2)  # 输出：[11 12 13]

# np.mean()
arr = np.array([1, 2, 3, 4, 5])
print(np.mean(arr))  # 输出：3.0

# np.median()
arr = np.array([1, 2, 3, 4, 5])
print(np.median(arr))  # 输出：3.0

# np.std()
arr = np.array([1, 2, 3, 4, 5])
print(np.std(arr))  # 输出：1.4142135623730951

#np.sum()
arr = np.array([1, 2, 3, 4, 5])
print(np.sum(arr))  # 输出：15
```

### 总结：
### Numpy 数组（ndarray） 是高效的同质数据容器，支持多维数据，并提供了许多强大的数值计算功能。
### 创建数组：可以通过 np.array(), np.zeros(), np.ones(), np.arange() 等方式创建数组。
### 数组索引与切片：通过整数索引和切片来访问和操作数组的元素。
### 基本运算：支持数组之间的加、减、乘、除运算以及广播机制。
### 常见函数：如 np.mean(), np.median(), np.std(), np.sum() 等常用的统计学函数，便于快速分析数据。

# 4.Pandas统计分析基础
## 4.1Pandas简介：介绍 Pandas 数据结构（Series 和 DataFrame）以及它的作用。
### Pandas 数据结构：Series 和 DataFrame
### 1️⃣	•	Series：一维数组，类似于 Python 中的列表，但它有一个索引，能更方便地进行数据查找和操作。
### 2️⃣	•	DataFrame：二维表格，类似于数据库中的表格或 Excel 工作簿的一个表单。它由多列组成，每列可以是不同的数据类型，且每一行有对应的索引。

```python
import pandas as pd

# 创建 Series
series = pd.Series([1, 2, 3, 4])
print(series)
# 输出：
# 0    1
# 1    2
# 2    3
# 3    4
# dtype: int64

# 创建 DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35]}
df = pd.DataFrame(data)
print(df)
# 输出：
#       Name  Age
# 0    Alice   25
# 1      Bob   30
# 2  Charlie   35
```

## 4.2DataFrame创建：如何从字典、列表、CSV 文件等创建 DataFrame

```python
# 可以通过字典来创建一个 DataFrame，其中字典的键是列名，值是对应列的列表。
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35]}
df = pd.DataFrame(data)
print(df)

# 如果我们有一个嵌套列表（列表的列表），可以将其直接转换成 DataFrame。
data = [['Alice', 25], ['Bob', 30], ['Charlie', 35]]
df = pd.DataFrame(data, columns=['Name', 'Age'])
print(df)

# 从 CSV 文件创建 DataFrame
df = pd.read_csv('data.csv')
print(df)
```

## 4.3数据选择与过滤：使用 .loc[] 和 .iloc[] 获取特定行列

```python
#使用 .loc[] 和 .iloc[] 获取特定行列
# .loc[]：基于 标签 来进行行列的选择，可以通过行的名称（标签）来选择数据。
#.iloc[]：基于 整数位置 来进行行列的选择，选择数据时使用的索引是整数类型。
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35]}
df = pd.DataFrame(data)

# 使用 .loc[] 通过标签选择行
print(df.loc[1])  # 输出第二行的数据
# 输出：
# Name     Bob
# Age       30
# Name: 1, dtype: object

# 使用 .iloc[] 通过整数位置选择行
print(df.iloc[0])  # 输出第一行的数据
# 输出：
# Name     Alice
# Age         25
# Name: 0, dtype: object

# 通过列名来选择特定的列
print(df['Name'])  # 输出 Name 列
# 输出：
# 0      Alice
# 1        Bob
# 2    Charlie
# Name: Name, dtype: object
```

## 4.4数据处理：如何处理缺失数据，使用 .dropna()、.fillna() 填充数据

```python
# 处理缺失数据：.dropna() 和 .fillna()
# .dropna()：删除包含缺失数据的行或列。
# .fillna()：填充缺失数据，可以使用常数值、均值、前后值等方式进行填充。
data = {'Name': ['Alice', 'Bob', None], 'Age': [25, None, 35]}
df = pd.DataFrame(data)

# 使用 .dropna() 删除包含缺失数据的行
df_dropped = df.dropna()
print(df_dropped)
# 输出：
#     Name   Age
# 0  Alice  25.0

# 使用 .fillna() 填充缺失数据
df_filled = df.fillna({'Name': 'Unknown', 'Age': 30})
print(df_filled)
# 输出：
#       Name   Age
# 0    Alice  25.0
# 1      Bob  30.0
# 2  Unknown  35.0
```

## 4.5基本统计：如何使用 .mean()、.sum()、.groupby() 等进行统计分析

```python
# .mean()：计算均值
df = pd.DataFrame({'Age': [25, 30, 35]})
print(df['Age'].mean())  # 输出：30.0

# .sum()：计算总和
df = pd.DataFrame({'Age': [25, 30, 35]})
print(df['Age'].sum())  # 输出：90

# .groupby()：分组
data = {'Name': ['Alice', 'Bob', 'Alice'], 'Age': [25, 30, 35]}
df = pd.DataFrame(data)

# 按 Name 列分组并计算 Age 的均值
grouped = df.groupby('Name')['Age'].mean()
print(grouped)
# 输出：
# Name
# Alice    30.0
# Bob      30.0
# Name: Age, dtype: float64
```

### 总结：
### Pandas 数据结构：Series 是一维数据，DataFrame 是二维数据表，能存储不同类型的数据。
### 创建 DataFrame：可以从字典、列表、CSV 文件等多种方式创建 DataFrame。
### 数据选择与过滤：使用 .loc[] 和 .iloc[] 分别基于标签和整数位置选择数据。
### 数据处理：通过 .dropna() 删除缺失数据，使用 .fillna() 填充缺失数据。
### 基本统计分析：通过 .mean() 计算均值，.sum() 计算总和，.groupby() 进行数据分组并应用聚合函数。

## 6.2数据准备：如何加载数据集（例如 load_iris()），以及如何使用 train_test_split() 切分训练集和测试集。

```python
# 一个常见的用于演示的机器学习数据集是 Iris 数据集。你可以使用 load_iris() 函数来加载这个数据集，它包含了 150 个花卉样本，数据包括花卉的萼片和花瓣的长度、宽度，以及花卉的类别。
from sklearn.datasets import load_iris

# 加载数据集
data = load_iris()
X = data.data  # 特征数据
y = data.target  # 目标变量
# load_iris() 返回一个包含数据和目标变量的字典对象。X 是一个二维数组（样本数 × 特征数），y 是一个一维数组，表示每个样本的类别。

# 训练集和测试集的切分
# 在机器学习中，我们通常将数据集分为训练集和测试集。训练集用于训练模型，测试集用于评估模型的性能。train_test_split() 是 scikit-learn 中用来切分数据集的函数。
from sklearn.model_selection import train_test_split

# 将数据集切分为训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
# test_size=0.3 表示测试集占 30%，训练集占 70%。
# random_state=42 用于确保每次切分的结果相同，方便复现。
```

## 6.3基础算法：介绍简单的机器学习算法（例如：线性回归、K-近邻算法）。

### 线性回归 (Linear Regression)
### 线性回归是一种简单的回归方法，用于预测数值型目标变量。它通过拟合数据点的最佳直线来建立输入特征与目标变量之间的线性关系。

```python
from sklearn.linear_model import LinearRegression

# 初始化模型
model = LinearRegression()

# 训练模型
model.fit(X_train, y_train)

# 预测
y_pred = model.predict(X_test)

# 输出预测结果
print(y_pred)
```

### K-近邻算法 (K-Nearest Neighbors, KNN)
### KNN 是一种简单的分类算法，它根据样本的邻近点来进行预测。对于分类问题，KNN 会选择测试点周围的 K 个最近邻点，选择其中出现次数最多的类别作为预测结果。

```python
from sklearn.neighbors import KNeighborsClassifier

# 初始化KNN分类器
knn = KNeighborsClassifier(n_neighbors=3)
# n_neighbors=3 表示选择 K=3 最近邻点。

# 训练模型
knn.fit(X_train, y_train)

# 预测
y_pred = knn.predict(X_test)

# 输出预测结果
print(y_pred)
```

## 6.4模型训练与评估：如何训练模型，如何评估模型性能（例如：准确度、混淆矩阵）。
### 在模型训练后，我们需要对模型进行评估，来检查它在未见数据上的表现。scikit-learn 提供了多种评估方法，包括准确度、混淆矩阵等。

### 评估模型性能

```python
# 1.准确度 (Accuracy)
# 准确度是衡量分类模型预测准确性的常见指标，表示预测正确的样本占总样本数的比例。
from sklearn.metrics import accuracy_score

# 计算准确度
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy}")

# 2.混淆矩阵 (Confusion Matrix)
# 混淆矩阵用于展示分类模型在各个类别上的表现，显示了真实标签与预测标签之间的关系。
from sklearn.metrics import confusion_matrix

# 计算混淆矩阵
cm = confusion_matrix(y_test, y_pred)
print(f"Confusion Matrix:\n{cm}")

# 3.其他评估指标
# 精确度、召回率、F1 分数等：这些可以通过 classification_report() 方法来计算。
from sklearn.metrics import classification_report

# 打印分类报告
report = classification_report(y_test, y_pred)
print(f"Classification Report:\n{report}")
```
