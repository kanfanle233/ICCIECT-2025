```python
# 一个常见的用于演示的机器学习数据集是 Iris 数据集。你可以使用 load_iris() 函数来加载这个数据集，它包含了 150 个花卉样本，数据包括花卉的萼片和花瓣的长度、宽度，以及花卉的类别。
from sklearn.datasets import load_iris

# 加载数据集
data = load_iris()
X = data.data  # 特征数据
y = data.target  # 目标变量
# load_iris() 返回一个包含数据和目标变量的字典对象。X 是一个二维数组（样本数 × 特征数），y 是一个一维数组，表示每个样本的类别。

# 训练集和测试集的切分
# 在机器学习中，我们通常将数据集分为训练集和测试集。训练集用于训练模型，测试集用于评估模型的性能。train_test_split() 是 scikit-learn 中用来切分数据集的函数。
from sklearn.model_selection import train_test_split

# 将数据集切分为训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
# test_size=0.3 表示测试集占 30%，训练集占 70%。
# random_state=42 用于确保每次切分的结果相同，方便复现。
```

```python
from sklearn.linear_model import LinearRegression

# 初始化模型
model = LinearRegression()

# 训练模型
model.fit(X_train, y_train)

# 预测
y_pred = model.predict(X_test)

# 输出预测结果
print(y_pred)
```
