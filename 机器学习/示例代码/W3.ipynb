## 1. 属性及其类型

### 在数据处理中，属性（Attribute） 指的是 用于描述数据对象的特征。不同的属性类型决定了如何处理数据。


```python
import pandas as pd

# 创建一个数据集
data = {
    "性别": ["男", "女", "男", "女"],  # 标称属性
    "学历": ["本科", "硕士", "博士", "本科"],  # 序数属性
    "年龄": [25, 30, 35, 40],  # 比率属性
    "温度": [36.5, 37.0, 38.2, 36.8]  # 间隔属性
}

df = pd.DataFrame(data)
print(df)
```

## 2. 数据的基本统计描述

### 数据分析的第一步通常是 查看数据的基本统计信息，例如 均值、中位数、方差、标准差、四分位数 等。

```python
# 均值（Mean） 数据的平均值 df.mean()
# 中位数（Median） 数据的中间值 df.median()
# 众数（Mode） 出现次数最多的值 df.mode()
# 标准差（Std） 反映数据离散程度 df.std()
# 方差（Variance） 反映数据波动大小 df.var()
# 最大值/最小值 数据的上下界 df.max() / df.min()
# 四分位数（Quantiles） 数据的分布范围 df.describe()
```

```python
import pandas as pd

# 创建 DataFrame
data = {
    "年龄": [25, 30, 35, 40, 45, 30, 35, 40, 50, 55],
    "收入": [3000, 3500, 4000, 5000, 6000, 3500, 4000, 4500, 7000, 8000],
    "评分": [3, 4, 5, 4, 3, 3, 4, 5, 5, 4]
}

df = pd.DataFrame(data)

# 输出数据
print("数据集：")
print(df)
```

```python
# 计算均值（Mean）
print("\n均值（Mean）：")
print(df.mean())

# 计算中位数（Median）
print("\n中位数（Median）：")
print(df.median())

# 计算众数（Mode）
print("\n众数（Mode）：")
print(df.mode())

# 计算标准差（Std）
print("\n标准差（Standard Deviation）：")
print(df.std())

# 计算方差（Variance）
print("\n方差（Variance）：")
print(df.var())

# 计算最大值（Max）和最小值（Min）
print("\n最大值（Max）：")
print(df.max())

print("\n最小值（Min）：")
print(df.min())

# 计算四分位数（Quantiles）
print("\n数据的描述性统计信息（df.describe()）：")
print(df.describe())
```

## 4. 数据对象的相似性度量

### 在许多任务（如聚类、推荐系统）中，我们需要衡量 数据对象之间的相似性，常见的度量方式包括 距离度量和相似度度量。


## 4.1 距离度量

## 4.1.1 欧氏距离（Euclidean Distance）

## 公式：

$$ d(A, B) = \sqrt{\sum (A_i - B_i)^2} $$

### 计算两个点之间的直线距离（最常见的距离度量）。
### 适用于 连续数值数据，如图像、物理空间坐标等。


```python
from scipy.spatial.distance import euclidean
import numpy as np

# 定义两个数据点
A = np.array([2, 3, 4])
B = np.array([5, 6, 7])

# 计算欧氏距离
distance = euclidean(A, B)
print(f"欧氏距离: {distance}")
```

## 4.1.2 曼哈顿距离（Manhattan Distance）

## 公式：

$$ d(A, B) = \sum |A_i - B_i| $$

### 计算两个点之间的 轴对齐 距离（如城市街区距离）。
### 适用于 网格化路径计算（如出租车行驶路线）。


```python
from scipy.spatial.distance import cityblock

# 计算曼哈顿距离
distance = cityblock(A, B)
print(f"曼哈顿距离: {distance}")
```

## 4.1.3 闵可夫斯基距离（Minkowski Distance）

##公式：

$$ d(A, B) = \left( \sum |A_i - B_i|^p \right)^{\frac{1}{p}} $$

### 欧氏距离和曼哈顿距离的推广，通过参数 p 控制计算方式：
### 当 p=2 → 欧氏距离。
### 当 p=1 → 曼哈顿距离。
### 当 p=\infty → 切比雪夫距离（Chebyshev Distance）。

```python
from scipy.spatial.distance import minkowski

# 计算闵可夫斯基距离，p=3
distance = minkowski(A, B, p=3)
print(f"闵可夫斯基距离 (p=3): {distance}")
```

## 4.2 相似度度量

## 4.2.1 余弦相似度（Cosine Similarity）

## 公式：

$$ \text{Similarity} (A, B) = \frac{A \cdot B}{||A|| \times ||B||} $$

### 计算 向量 A 和 B 之间的夹角，范围 [-1,1]，用于 文本分析、推荐系统。
### 1 代表完全相似，0 代表无相似性，-1 代表完全相反。

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
from IPython.display import display, Math

# 定义两个向量
A = np.array([[1, 2, 3]])
B = np.array([[2, 3, 4]])

# 计算余弦相似度
cos_sim = cosine_similarity(A, B)

# 显示公式
display(Math(r"\text{Similarity} (A, B) = \frac{A \cdot B}{||A|| \times ||B||}"))

print(f"余弦相似度: {cos_sim[0][0]}")
```

## 4.2.2 Jaccard 相似度（Jaccard Similarity）

##公式：

$$ J(A, B) = \frac{|A \cap B|}{|A \cup B|} $$

### 计算 两个集合的交集与并集的比值，适用于 集合数据、推荐系统。
### 范围 [0,1]，值越大，表示相似度越高。

```python
from sklearn.metrics import jaccard_score

# 定义两个二进制集合（Jaccard 适用于 0/1 数据）
A = np.array([1, 1, 0, 1, 0, 1, 0])
B = np.array([1, 0, 1, 1, 0, 1, 1])

# 计算 Jaccard 相似度
jaccard_sim = jaccard_score(A, B)

# 显示公式
display(Math(r"J(A, B) = \frac{|A \cap B|}{|A \cup B|}"))

print(f"Jaccard 相似度: {jaccard_sim}")
```

## 4.2.3 皮尔逊相关系数（Pearson Correlation Coefficient）

##公式：

$$ r = \frac{\sum (X_i - \bar{X}) (Y_i - \bar{Y})}{\sqrt{\sum (X_i - \bar{X})^2} \times \sqrt{\sum (Y_i - \bar{Y})^2}} $$

### 衡量 两个变量的线性相关性，适用于 数值数据、回归分析。
### 范围 [-1,1]：
### r > 0 → 正相关
### r = 0 → 无相关
### r < 0 → 负相关

```python
from scipy.stats import pearsonr

# 定义两个变量
X = np.array([10, 20, 30, 40, 50])
Y = np.array([5, 15, 25, 35, 45])

# 计算皮尔逊相关系数
pearson_corr, _ = pearsonr(X, Y)

# 显示公式
display(Math(r"r = \frac{\sum (X_i - \bar{X}) (Y_i - \bar{Y})}{\sqrt{\sum (X_i - \bar{X})^2} \times \sqrt{\sum (Y_i - \bar{Y})^2}}"))

print(f"皮尔逊相关系数: {pearson_corr}")
```
