```python
# 使用TensorFlow的梯度下降法求解方程: x/2 + 30 = 4/5 * x
# 思路：将方程求解转化为优化问题，最小化 (x/2 + 30 - 0.8*x)^2

import tensorflow as tf

# 1. 定义可训练参数 x，初始值设为1.0
x = tf.Variable(1.0, dtype=tf.float32)

# 2. 使用Adam优化器，学习率设为0.1（Adam自适应学习率，收敛更稳定）
optimizer = tf.keras.optimizers.Adam(learning_rate=0.1)

# 3. 训练循环：500步梯度下降
for step in range(500):
    with tf.GradientTape() as tape:          # 记录计算图，用于自动求导
        lhs = x / 2 + 30                     # 方程左边：x/2 + 30
        rhs = 0.8 * x                        # 方程右边：4/5 * x = 0.8x
        loss = tf.square(lhs - rhs)          # 损失函数：差值的平方
    grad = tape.gradient(loss, x)            # 计算损失对x的梯度
    optimizer.apply_gradients([(grad, x)])    # 根据梯度更新x的值
    
    if step % 50 == 0:
        print(f"Step {step:03d} -> x = {x.numpy():.6f}, loss = {loss.numpy():.6f}")

# 4. 最终结果
print(f"\n最终收敛结果：x = {x.numpy():.6f}")
print(f"方程残差：{(x.numpy()/2 + 30 - 0.8*x.numpy()):.6e}")
```
